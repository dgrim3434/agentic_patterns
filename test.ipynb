```python
import os
from anthropic import Anthropic
from dotenv import load_dotenv
```

```python
c = Anthropic(
    api_key = os.environ.get("ANTHROPIC_API_KEY")
)
```

```python
r = c.messages.create(
    model="claude-sonnet-5-5",
    max_tokens=100,
    messages=[{'role': 'user', 'content': "say hi"}]
)
```

```python
def create_output(message, client, system):
    
    resp = client.messages.create(
        max_tokens=8000,
        messages = message,
        model = "claude-haiku-5-5",
        system=system,
    )
    text = " ".join([c.text for c in resp.content if c.type == "text"])
    
    return text, resp
```

```python
system = (
            "You are a Python programmer tasked with generating high quality Python code. "
            "Your task is to generate the best content possible for the user's request. "
            "If the user provides critique, respond with a revised version of your previous attempt."
)

generation_chat_history = [{
    'role': 'user',
    'content': "Generate a Python implementation of the Merge Sort Algorithm"
}]

message = create_output(generation_chat_history, c, system)
```

```python
generation_chat_history.append({'role': 'assistant', 'content': message[0]})
```

```python
reflection_system = ("You are Andrej Karpathy, an experienced computer scientist. "
                     "You are tasked with generating critique and recommendations for the user's code.")

reflection_message = [{'role': 'user', 'content': message[0]}]

crit_resp = create_output(reflection_message, c, reflection_system)
```
